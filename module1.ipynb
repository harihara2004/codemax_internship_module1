```python

# PYTHON LIST & DICTIONARY PRACTICE


#  STORE DATA
students = ["Hari", "Arun", "Priya", "Kumar"]

student = {
    "name": "Hari",
    "age": 22,
    "course": "MSc Data Science",
    "cgpa": 7.77,
    "skills": ["Python", "SQL", "Machine Learning"]
}

print("ORIGINAL DATA ")
print("Students:", students)
print("Student:", student)


# ACCESS LIST DATA
print("\n ACCESS LIST ")

print("First student:", students[0])
print("Second student:", students[1])
print("Last student:", students[-1])
print("First two students:", students[0:2])


# ACCESS DICTIONARY DATA
print("\n ACCESS DICTIONARY ")

print("Name:", student["name"])
print("Age:", student["age"])
print("Course:", student["course"])
print("CGPA:", student["cgpa"])

# Access nested list inside dictionary
print("First skill:", student["skills"][0])


# ADD DATA
print("\n ADD DATA ")

# Add a new student to the list
students.append("Vijay")

# Add a new skill to the nested list
student["skills"].append("Pandas")

# Add a new key-value pair
student["city"] = "Chennai"

print("After adding:", students)
print("Skills:", student["skills"])
print("City:", student["city"])


#  UPDATE DATA
print("\n UPDATE DATA ")

# Update list value
students[1] = "Rahul"

# Update dictionary value
student["age"] = 23
student["cgpa"] = 8.25

print("Updated students:", students)
print("Updated age:", student["age"])
print("Updated CGPA:", student["cgpa"])


#  REMOVE DATA
print("\n REMOVE DATA ")

# Remove a student from the list
students.remove("Priya")

# Remove a key-value pair
student.pop("city")

print("After removing student:", students)
print("After removing city:", student)


#  LOOP THROUGH LIST
print("\n LIST LOOP ")

for name in students:
    print("Student name:", name)


#  LOOP THROUGH DICTIONARY
print("\n  DICTIONARY LOOP ")

for key, value in student.items():
    print(key, ":", value)


#  CHECK DATA
print("\n CHECK DATA ")

if "Hari" in students:
    print("Hari is available in the list")
else:
    print("Hari is not available")

if "name" in student:
    print("Name key is available")
else:
    print("Name key is not available")


# . FINAL DATA
print("\n FINAL DATA ")

print("Final Students:", students)
print("Final Student Details:", student)
```

```python
# File Handling in Python

#  WRITE to a file
# "w" creates a new file or overwrites an existing file.
with open("student.txt", "w") as file:
    file.write("Name: Hariharan\n")
    file.write("Course: MSc Applied Data Science\n")
    file.write("Skill: Python\n")

print("File written successfully!")


#  READ the file
# "r" reads the contents of the file.
with open("student.txt", "r") as file:
    data = file.read()

print("\nFile contents:")
print(data)


#  APPEND to the file
# "a" adds new content without deleting existing content.
with open("student.txt", "a") as file:
    file.write("Goal: Machine Learning Engineer\n")

print("New content added successfully!")


#  READ again to see the updated file
with open("student.txt", "r") as file:
    data = file.read()

print("\nUpdated file contents:")
print(data)
```
