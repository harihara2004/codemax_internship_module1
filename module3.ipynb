```python
# Import required libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Set style for better visual aesthetics
sns.set_theme(style="whitegrid")
```

```python
# Create a sample dataset for demonstration
np.random.seed(42)
data = {
    "Day": [f"Day {i}" for i in range(1, 11)],
    "Study_Hours": np.random.randint(2, 8, size=10),
    "Scores": np.random.randint(60, 95, size=10),
    "Category": np.random.choice(["Coding", "Reading", "Practice"], size=10),
}
df = pd.DataFrame(data)
df
```

```python
# 1. Line Chart: Trends over Days
plt.figure(figsize=(10, 5))
plt.plot(
    df["Day"],
    df["Scores"],
    marker="o",
    color="b",
    linestyle="-",
    linewidth=2,
    markersize=8,
)
plt.title("Performance Trend Over Days", fontsize=14, fontweight="bold")
plt.xlabel("Days", fontsize=12)
plt.ylabel("Scores", fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Create a sample dataset for demonstration for self-containment
np.random.seed(42)
data = {
    "Day": [f"Day {i}" for i in range(1, 11)],
    "Study_Hours": np.random.randint(2, 8, size=10),
    "Scores": np.random.randint(60, 95, size=10),
    "Category": np.random.choice(["Coding", "Reading", "Practice"], size=10),
}
df = pd.DataFrame(data)

# 2. Bar Chart: Study Hours per Day
plt.figure(figsize=(10, 5))
plt.bar(df["Day"], df["Study_Hours"], color="skyblue", edgecolor="black")
plt.title("Daily Study Hours", fontsize=14, fontweight="bold")
plt.xlabel("Days", fontsize=12)
plt.ylabel("Hours", fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Create a sample dataset for demonstration for self-containment
np.random.seed(42)
data = {
    "Day": [f"Day {i}" for i in range(1, 11)],
    "Study_Hours": np.random.randint(2, 8, size=10),
    "Scores": np.random.randint(60, 95, size=10),
    "Category": np.random.choice(["Coding", "Reading", "Practice"], size=10),
}
df = pd.DataFrame(data)

# 3. Pie Chart: Category Distribution
category_counts = df["Category"].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(
    category_counts,
    labels=category_counts.index,
    autopct="%1.1f%%",
    startangle=140,
    colors=sns.color_palette("pastel"),
)
plt.title("Distribution of Activities", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
# 4. Histogram: Distribution of Scores
plt.figure(figsize=(8, 5))
plt.hist(df["Scores"], bins=5, color="purple", edgecolor="black", alpha=0.7)
plt.title("Score Distribution Histogram", fontsize=14, fontweight="bold")
plt.ylabel("Frequency", fontsize=12)
plt.xlabel("Scores", fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# 5. Scatter Plot with Seaborn: Study Hours vs Scores
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Scores",
    hue="Category",
    s=100,
    palette="deep",
)
plt.title("Relationship Between Study Hours and Scores", fontsize=14, fontweight="bold")
plt.xlabel("Study Hours", fontsize=12)
plt.ylabel("Scores", fontsize=12)
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```
